# Cluster 2

```python
def predict_cluster2(df_new):
    with open("models/cluster2_model.pkl", "rb") as f:
        model = pickle.load(f)
    with open("models/cluster2_threshold.pkl", "rb") as f:
        thr = pickle.load(f)
    with open("models/cluster2_features.pkl", "rb") as f:
        cols = pickle.load(f)
    X_new = df_new[cols].values
    probs = model.predict_proba(X_new)[:, 1]
    return (probs >= thr).astype(int)
```

# Cluster #4

```python
with open('group4processing.pkl', 'rb') as f:
    loaded_data = pickle.load(f)
loaded_data
```

```python
df4 = X[X['Cluster'] == 4]
df4
```

```python
def preprocess_group4(X):
    X = X.drop(['Net Income Flag', 'Liability-Assets Flag', 'Cluster'], axis=1)
    data = loaded_data['selector'].transform(X)
    selected_features = X.columns[loaded_data['selector'].get_support()].tolist()
    df = pd.DataFrame(data, columns=selected_features)

    data1 = loaded_data['normal1'].transform(df)
    df1 = pd.DataFrame(data1, columns=selected_features)

    data2 = loaded_data['pca'].transform(df1)
    pc_df = pd.DataFrame(data2, columns=[f'Feature Number {i+1}' for i in range(data2.shape[1])])

    data3 = loaded_data['normal2'].transform(pc_df)
    pc_df = pd.DataFrame(data3, columns=pc_df.columns)
    return pc_df

processed_data = preprocess_group4(df4)
processed_data
```

```python
def predict4(X):
    processed_data = preprocess_group4(X)
    return loaded_data['subgroup_model'].predict(processed_data)
```

# Cluster 0

```python
with open('subgroup_0_model.pkl', 'rb') as f:
    pipeline = pickle.load(f)
pipeline.steps

df_0 = X[X['Cluster'] == 0]
def process_subgroup0(pipeline, df):
    df1 = df.drop('Cluster', axis=1)
    scaler = pipeline.named_steps['scale']
    selector = pipeline.named_steps['select']
    df0_scaled = scaler.transform(df1)
    df0_reduced = selector.transform(df0_scaled)
    return df0_reduced

def predict0(pipeline, df, stacking_model):
    X = process_subgroup0(pipeline, df)
    predictions = stacking_model.predict(X)
    print("count 0:", np.sum(predictions==0))
    print("count 1:", np.sum(predictions==1))
    return predictions
```

```python
bankrupt = []
for idx in X.index:
    row = X.loc[[idx]] 
    cl = row['Cluster'].iat[0]
    if cl == 0:
        bankrupt.append(predict0(pipeline, row, pipeline.named_steps['model']))
        #print(predict0(pipeline, row, pipeline.named_steps['model']))
    elif cl == 1:
        bankrupt.append(predict_my_subgroup(row)[0])
        #print(predict_my_subgroup(row)[0])
    elif cl == 2:
        bankrupt.append(predict_cluster2(row)[0])
    elif cl == 3:
        bankrupt.append(0)
    elif cl == 4:
        bankrupt.append(predict4(row)[0])
        

answer = pd.DataFrame({
    'Index': index,
    'Bankrupt?': bankrupt
})
answer
```
